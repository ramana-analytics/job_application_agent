```python
import requests
from bs4 import BeautifulSoup

def extract_text_from_url(url):
    # Step 1: Fetch the page
    response = requests.get(url, timeout=10)
    response.raise_for_status()  # raises error for bad status codes

    # Step 2: Parse HTML
    soup = BeautifulSoup(response.text, "html.parser")

    # Step 3: Extract visible text
    # Remove scripts, styles, etc.
    for tag in soup(["script", "style", "noscript"]):
        tag.extract()

    text = soup.get_text(separator="\n")
    
    # Step 4: Clean whitespace
    lines = [line.strip() for line in text.splitlines() if line.strip()]
    return "\n".join(lines)

# Example usage
url = "https://recruiting2.ultipro.com/MON1026MONOH/JobBoard/bcaf2db0-50c8-4e2f-9d65-e8e00c882a25/OpportunityDetail?opportunityId=0003152c-8149-40fb-bc50-f58a154d995e&source=LinkedIn&utm_source=LINKEDIN"
content = extract_text_from_url(url)
print(content)

```
